```python
import pandas as pd
import numpy as np

listings = pd.read_csv("data/listings.csv")
```

```python
listings["property_type"].value_counts(normalize=True).round(3).head(5)
```

```python
listings["neighbourhood_cleansed"].value_counts()
```

```python
listings.groupby("room_type")["beds"].mean().round(2)
```

```python
listings.groupby("room_type")["beds"].agg(["count", "mean"]).round(2)
```

```python
listings.groupby("neighbourhood_cleansed")["number_of_reviews"].agg(["count", "mean", "median"]).round(2).sort_values("median")
```

```python
listings.groupby("room_type").agg(
    n_listings=("id", "count"),
    avg_beds=("beds", "mean"),
    max_reviews=("number_of_reviews", "max"),
).round(2)
```

```python
pd.crosstab(listings["neighbourhood_cleansed"], listings["host_is_superhost"], normalize="index").round(3)
```

```python
listings["has_reviews"] = listings["number_of_reviews"] > 0
listings["has_reviews"]
```

```python
listings["popular"] = np.where(listings["number_of_reviews"] > 50, "popular", "not popular")
listings["popular"]
```

```python
bins = [0, 1, 6, 29, float("inf")]
labels = ["1 night", "2-6 nights", "7-29 nights", "30+ nights"]

listings["stay"] = pd.cut(listings["minimum_nights"], bins=bins, labels=labels)

print(listings["stay"].value_counts().sort_index())
print("Listings with no group:", listings["stay"].isna().sum())
```
